## StellarMasses table

According to http://www.gama-survey.org/dr3/schema/dmu.php?id=9:

The results given in table StellarMasses have been derived from
matched aperture photometry, i.e. the r-defined AUTO photometry from
ApMatchedCat (Hill et al. 2011; Liske et al. 2015; Driver et
al. 2015). In StellarMassesLambdar results are also given for the
photometry from LambdarCat derived using the code LAMBDAR (Wright
et al. 2016). This code attempts to derive consistent and deblended
photometry across all bands. (Note that the convention in both of
these photometry catalogues is to refer to the SDSS z-band and the
VISTA Z-band filters as 'z' and 'X', respectively. This convention is
perpetuated in this DMU). Neither the AUTO nor the LAMBDAR photometry
purports to necessarily represent the total flux in any particular
band.

AN APERTURE CORRECTION IS THEREFORE REQUIRED for integrated quantities
(i.e. stellar mass or luminosity), in order to account for mass/flux
that falls beyond the finite AUTO aperture used for the SEDs. The
column 'fluxscale' is provided for this purpose. The quantity
'fluxscale' is the (linear) ratio between the r-band (AUTO/LAMBDAR)
aperture flux and the total r-band flux inferred from fitting a Sersic
profile (truncated at 10 Re, i.e. 'GAL_MAG_10RE_R', taken from
SersicCatSDSS).

THIS CORRECTION IS NOT APPLIED BY DEFAULT for two reasons. First, not
all objects have a Sersic-fit magnitude owing to failures in either
the SExtractor detection/segmentation or the GALFIT Sersic fitting
procedures (see Kelvin et al. 2012). These objects can be identified
as those with 'fluxscale' = -9999. Second, particularly for the
fainter galaxies with higher (but unquantified) uncertainties in the
Sersic-fit magnitudes, the flux scaling implied by the Sersic
magnitudes is not necessarily physically sensible. For many galaxies,
the Sersic magnitude is less than the aperture magnitude, which is
logically inconsistent. Further, for some galaxies, the implied flux
scaling leads to masses that are unrealistically high (log M* >> 13).

THE USER IS THEREFORE TRUSTED TO APPLY THESE CORRECTIONS, but with the
understanding that some care is required. The recommended mass to use
is thus:
   log10 M*,total = logmstar + log10(fluxscale) - 2 log10(h/0.7), 
and the recommended absolute magnitude in the X-band is:
   M_X,total = absmag_X - 2.5 log10(fluxscale) + 5 log10(h/0.7).
   
Users should check to ensure that this makes sense (e.g. check that
'fluxscale' is greater than or equal to unity, if only approximately,
and that the magnitude of the correction is not a factor of many) for
their specific object(s) of interest. Note, too, that this correction
introduces an additional source of uncertainty in the total
luminosities and masses that is not quantified in these tables.

```python
from astroquery.gama import GAMA
from astropy import coordinates
import astropy.units as u
```

```python
result = GAMA.query_sql("""SELECT WISECat.desig_allwise AS 'wiseX', StellarMasses.logmstar AS 'log_sm',
                            StellarMasses.dellogmstar AS 'd_log_sm', StellarMasses.fluxscale AS 'fluxscale'
                            FROM WISECat INNER JOIN StellarMasses 
                            ON StellarMasses.cataid = WISECat.cataid""")
print(result)
```

```python
type(result)
```

```python
result.write( 'GAMA_sm.txt', format='csv', overwrite=True )
```

## StellarMassesLambdar table

```python
result = GAMA.query_sql("""SELECT WISECat.desig_allwise AS 'wiseX', StellarMassesLambdar.logmstar AS 'log_sm',
                            StellarMassesLambdar.dellogmstar AS 'd_log_sm', 
                            StellarMassesLambdar.fluxscale AS 'fluxscale'
                            FROM WISECat INNER JOIN StellarMassesLambdar 
                            ON StellarMassesLambdar.cataid = WISECat.cataid""")
print(result)
```

```python
result.write( 'GAMA_sm_lambdar.txt', format='csv', overwrite=True )
```

## Magphys table

From http://www.gama-survey.org/dr3/schema/table.php?id=82:

Best fit total mass of stars ever formed; integral of the star formation rate

```python
result = GAMA.query_sql("""SELECT WISECat.desig_allwise AS 'wiseX',
                            MagPhys.mass_stellar_best_fit AS 'sm'
                            FROM WISECat INNER JOIN MagPhys
                            ON MagPhys.cataid = WISECat.cataid""")
print(result)
```

```python
result.write( 'GAMA_sm_magphys.txt', format='csv', overwrite=True )
```
