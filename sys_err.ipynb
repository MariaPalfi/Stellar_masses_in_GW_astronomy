```python
# importing useful packages
import numpy as np
import pandas as pd
```

```python
# reading the file to 'data' pandas dataframe
check = pd.read_csv( '../check_sm.txt', delimiter = '\t', low_memory = False )
check.head()
```

```python
sm_std = []
for i in range( 0, len(check) ):
    sm_std.append( np.std( [ 10**check.log_gama_sm[i], 10**check.log_gama_sm_lambdar[i], 10**check.log_gama_sm_magphys[i] ] ) )
```

```python
sm_mean = []
for i in range( 0, len(check) ):
    sm_mean.append( np.mean( [ 10**check.log_gama_sm[i], 10**check.log_gama_sm_lambdar[i], 10**check.log_gama_sm_magphys[i] ] ) )
```

```python
 np.array(sm_std)/np.array(sm_mean) 
```

```python
np.log10(sm_std[:10])
```

```python
np.mean( np.array(sm_std)/np.array(sm_mean) )
```
