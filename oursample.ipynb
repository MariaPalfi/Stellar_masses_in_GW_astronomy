```python
# importing useful packages
import numpy as np
import pandas as pd
import copy
```

Reading and merging data 

```python
# reading the file to 'data' pandas dataframe
data = pd.read_csv( '../sm_both.txt', delimiter = '\t', low_memory = False )
```

```python
gama = pd.read_csv( '../GAMA_sm.txt' )
gama = gama[ gama.fluxscale != -9999 ]
gama['log_gama_sm'] = gama.log_sm + np.log10( gama.fluxscale ) - 2 * np.log10( 67.66/(100*0.7) )
gama = copy.deepcopy( gama[gama.log_gama_sm > 5])

check = pd.merge(data, gama, on=['wiseX'], how='inner')
```

```python
gama = pd.read_csv( '../GAMA_sm_lambdar.txt' )
gama = gama[ gama.fluxscale != -9999 ]
gama['log_gama_sm'] = gama.log_sm + np.log10( gama.fluxscale ) - 2 * np.log10( 67.66/(100*0.7) )
gama = copy.deepcopy( gama[gama.log_gama_sm > 5])

check = pd.merge(check, gama, on=['wiseX'], how='inner', suffixes = ( '', '_lambdar') )
```

```python
gama = pd.read_csv( '../GAMA_sm_magphys.txt' )
gama['log_gama_sm'] = np.log10(gama.sm)
gama = copy.deepcopy( gama[gama.log_gama_sm > 5])

check = pd.merge(check, gama, on=['wiseX'], how='inner', suffixes = ( '', '_magphys'))
```
